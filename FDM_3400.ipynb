### Fractional Dark Matter:
$$ FDM = m_{dyn} - (m_{stellar} + m_{gas}) $$
$$  $$
  
### Dynamical Mass
$$ M_{dyn} = \frac{C \sigma^2 R_e}{G} $$
$$  $$
C $\approx$ 5, $\sigma =$ velocity dispersion(given), $R_e$ = half-life radius(given), G = gravitational constant  
  
### Stellar Mass
$$ m_{stellar} = K * L$$
$$  $$
K is the mass-to-light ratio
#### Luminosity
$$ \frac{L}{L_{sun}} = 10^{0.4(M_{sun} - M)} $$

```python
# Constants
K = 5 # Solar mass / Solar Luminosity
c = co.c # Speed of light
G = co.G # Gravtiational Constants
L_sun = co.L_sun # Luminosity of the Sun
M_sun = co.M_sun # Mass of the Sun
M_bol_sun = -2.5 * np.log10(L_sun.value / co.L_bol0.value) # Absolute bolometric magnitude of the Sun
H = 70. * u.km/u.s/u.Mpc # Hubble's constant
z = 0.04027193 # Redshift (WILL NEED TO CHANGE WHEN WE PULL DATA)
cosm = cosmo.FlatLambdaCDM(H0=H, Om0=0.3) # Using FlatLambdaCDM for calculations
```

```python
# Distance
d = cosm.luminosity_distance(z)
```

```python
dist_modulus = cosm.distmod(z)  # Returns distance modulus in magnitudes
apparent_mag_g = 15.298
absolute_mag_g = apparent_mag_g - dist_modulus.value
```

```python
lum = 10**(0.4*(M_bol_sun - absolute_mag_g))
```

```python
M_s_sol = lum * K
M_s = M_s_sol * M_sun
```
