## Math Equations 
### Fractional Dark Matter:
$$ FDM = m_{dyn} - (m_{stellar} + m_{gas}) $$
$$  $$
  
### Dynamical Mass
$$ M_{dyn} = \frac{C \sigma^2 R_e}{G} $$
$$  $$
C $\approx$ 5, $\sigma =$ velocity dispersion(given), $R_e$ = half-life radius(given), G = gravitational constant  
  
### Stellar Mass
$$ m_{stellar} = K * L$$
$$  $$
K is the mass-to-light ratio
#### Luminosity
$$ \frac{L}{L_{sun}} = 10^{0.4(M_{sun} - M)} $$
$$  $$
Where M represents absolute magnitude
